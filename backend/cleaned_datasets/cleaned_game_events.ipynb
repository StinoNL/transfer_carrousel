```python
import pandas as pd

# loading game_events.csv
df = pd.read_csv('kaggle raw data/game_events.csv')

# checking basic info
print(df.shape)
df.head()
```

```python
df['date'] = pd.to_datetime(df['date'], errors='coerce')
```

```python
df['description'] = df['description'].astype(str).str.strip().str.lstrip(',').str.strip()
```

```python
df['player_in_id'] = df['player_in_id'].fillna(-1).astype(int)
df['player_assist_id'] = df['player_assist_id'].fillna(-1).astype(int)
```

```python
df['minute'] = pd.to_numeric(df['minute'], errors='coerce').fillna(-1)
```

```python
df['type'] = df['type'].str.strip().str.title()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df['is_goal'] = df['type'] == 'Goals'
df['is_card'] = df['type'] == 'Cards'
df['is_substitution'] = df['type'] == 'Substitutions'
```

```python
df['minute_bucket'] = pd.cut(df['minute'], bins=[-1, 15, 30, 45, 60, 75, 90, 120],
                             labels=['0–15', '16–30', '31–45', '46–60', '61–75', '76–90', '90+'])
```

```python
df['has_assist'] = df['player_assist_id'] != -1
```

```python
df['card_type'] = df['description'].apply(
    lambda x: 'Yellow' if 'Yellow' in x else 'Red' if 'Red' in x else 'Unknown'
)
```

```python
df['is_header'] = df['description'].str.contains('Header', case=False, na=False)
df['is_own_goal'] = df['description'].str.contains('Own Goal', case=False, na=False)
```

```python
print(df.dtypes)
```

```python
print(df['minute_bucket'].value_counts())
print(df['has_assist'].value_counts())
```

```python
import numpy as np

df['player_in_id'] = df['player_in_id'].replace(-1, np.nan)
df['player_assist_id'] = df['player_assist_id'].replace(-1, np.nan)
df['minute'] = df['minute'].replace(-1, np.nan)
```
