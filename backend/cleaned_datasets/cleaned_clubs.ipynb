```python
import pandas as pd

df = pd.read_csv('kaggle raw data/clubs.csv')
print(df.shape)
df.head()
```

```python
df.isna().sum().sort_values(ascending=False)
```

```python
# filling missing percentages and average_age with median
df['foreigners_percentage'] = df['foreigners_percentage'].fillna(df['foreigners_percentage'].median())
df['average_age'] = df['average_age'].fillna(df['average_age'].median())
```

```python
def clean_transfer_value(val):
    if pd.isna(val): return None
    val = val.replace('€', '').replace('m', 'e6').replace('k', 'e3').replace('+', '').replace(',', '')
    try:
        return eval(val)
    except:
        return None

df['net_transfer_record'] = df['net_transfer_record'].apply(clean_transfer_value)
```

```python
df['total_market_value'] = df['total_market_value'].fillna(0)

df['coach_name'] = df['coach_name'].fillna('Unknown')
```

```python
def categorize_net_transfer(val):
    if val < -10_000_000:
        return 'spender'
    elif val > 10_000_000:
        return 'saver'
    else:
        return 'balanced'

df['net_transfer_category'] = df['net_transfer_record'].apply(categorize_net_transfer)
```

```python
print(df.dtypes)
```

```python
df.to_csv('cleaned_clubs.csv', index=False)
```
